## 3. Un conto vero, per essere sicuri

Se questa cella stampa un tempo, il calcolo funziona davvero. Su GPU dura una frazione
di secondo, su CPU qualche secondo: vanno bene entrambe.

```python
import time

import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
a = torch.randn(2000, 2000, device=device)

start = time.perf_counter()
b = a @ a
if device == "cuda":
    torch.cuda.synchronize()
durata = time.perf_counter() - start

print(f"Moltiplicazione 2000x2000 su {device}: {durata:.3f} secondi")
print("Risultato finito, tutto a posto.")
```
